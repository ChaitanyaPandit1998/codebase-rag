# Qdrant Hello World

A simple example demonstrating basic Qdrant vector database operations:
- Create a collection
- Insert vectors (points)
- Search for similar vectors

```python
from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams, PointStruct
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
api_key = os.getenv("QDRANT_API_KEY")
url = os.getenv("QDRANT_URL")
```

```python
qdrant_client = QdrantClient(
    url=url,
    api_key=api_key
)

print(qdrant_client.get_collections())
```

```python
# Create a collection with 4-dimensional vectors using cosine similarity
qdrant_client.create_collection(
    collection_name="hello_world",
    vectors_config=VectorParams(size=4, distance=Distance.COSINE),
)
print("Collection 'hello_world' created")
```

```python
# Insert some points with vectors and payloads
points = [
    PointStruct(id=1, vector=[0.1, 0.9, 0.1, 0.1], payload={"text": "apple", "category": "fruit"}),
    PointStruct(id=2, vector=[0.9, 0.1, 0.1, 0.1], payload={"text": "car", "category": "vehicle"}),
    PointStruct(id=3, vector=[0.1, 0.8, 0.2, 0.1], payload={"text": "banana", "category": "fruit"}),
    PointStruct(id=4, vector=[0.8, 0.2, 0.1, 0.1], payload={"text": "truck", "category": "vehicle"}),
]

qdrant_client.upsert(collection_name="hello_world", points=points)
print(f"Inserted {len(points)} points")
```

```python
# Search for vectors similar to "apple" [0.1, 0.9, 0.1, 0.1]
results = qdrant_client.query_points(
    collection_name="hello_world",
    query=[0.1, 0.85, 0.15, 0.1],
    limit=3,
)

print("Search results (closest to fruit-like vector):")
for point in results.points:
    print(f"  id={point.id}, score={point.score:.4f}, payload={point.payload}")
```

```python
# Get collection info
info = qdrant_client.get_collection("hello_world")
print(f"Collection: {info.config.params.vectors.size}D vectors, {info.points_count} points")
```
